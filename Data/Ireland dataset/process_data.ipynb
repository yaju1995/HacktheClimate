```python
import pandas as pd
```

```python
Year = 2026
```

#### Loading forecasted data from https://www.smartgriddashboard.com/roi/wind/?duration=month&datefrom=28-Dec-2025&dateto=26-Jan-2026

```python
df = pd.read_csv(f'./EirGrid_data/{Year}/Wind_generation_ireland.csv')
df.columns = df.columns.str.strip() 
df["DATE & TIME"] = pd.to_datetime(
        df["DATE & TIME"],
        dayfirst=True,
        errors="raise",        # fail loudly if format is wrong
        format="%d/%m/%Y %H:%M",
        utc=True
    )
df = df.set_index("DATE & TIME")
```

```python
print(df.columns)
```

```python
# Hourly average
wind_hourly = df.resample("h").agg({"ACTUAL WIND(MW)": "mean","FORECAST WIND(MW)": "mean"})
```

#### Loading weather data from :
https://www.met.ie/climate/available-data/historical-data

```python
weather_hourly= pd.read_csv(f'Met-Eireann_data/{Year}/AggregateData_weather.csv')
weather_hourly["date"] = pd.to_datetime(
        weather_hourly["date"],
        format="%d/%m/%Y %H:%M",
        dayfirst=True,
        errors="raise",
        utc=True
    )

weather_hourly = weather_hourly.set_index("date")
```

```python
weather_hourly
```

#### Loading Entso-e data set

```python
entsoe_actual_generation = pd.read_csv(f'./Entso-e_data/{Year}/actual_generation_ENTSOE.csv')                            
entsoe_forecasted_wind_solar = pd.read_csv(f'./Entso-e_data/{Year}/wind_solar_forecast_ENTSOE.csv')
entsoe_load = pd.read_csv(f'./Entso-e_data/{Year}/load_ENTSOE.csv')
entsoe_load_forecast = pd.read_csv(f'./Entso-e_data/{Year}/load_forecast_ENTSOE.csv')
```

```python
entsoe_actual_generation = entsoe_actual_generation.set_index("DateTime")
entsoe_actual_generation.index = pd.to_datetime(entsoe_actual_generation.index, utc=True, errors="coerce")

entsoe_forecasted_wind_solar = entsoe_forecasted_wind_solar.set_index("DateTime")
entsoe_forecasted_wind_solar.index = pd.to_datetime(entsoe_forecasted_wind_solar.index, utc=True, errors="coerce")

entsoe_load = entsoe_load.set_index("DateTime")
entsoe_load.index = pd.to_datetime(entsoe_load.index, utc=True, errors="coerce")

entsoe_load_forecast = entsoe_load_forecast.set_index("DateTime")
entsoe_load_forecast.index = pd.to_datetime(entsoe_load_forecast.index, utc=True, errors="coerce")
```

```python
entsoe_hourly_actual_generation = entsoe_actual_generation.resample("h").mean()
entsoe_hourly_forecasted_wind_solar = entsoe_forecasted_wind_solar.resample("h").mean()
entsoe_hourly_load = entsoe_load.resample("h").mean()
entsoe_hourly_load_forecast = entsoe_load_forecast.resample("h").mean()
```

#### Merging weather data with generation data

```python
def merge_dataframes_wrt_time(df1, df2, name1="df1", name2="df2"):
    # Ensure both indexes are DatetimeIndex and tz-aware
    df1.index = pd.to_datetime(df1.index, utc=True)
    df2.index = pd.to_datetime(df2.index, utc=True)

    # Extract timestamp sets
    ts1 = set(df1.index)
    ts2 = set(df2.index)

    # Compute missing timestamps
    missing_in_df1 = sorted(ts2 - ts1)
    missing_in_df2 = sorted(ts1 - ts2)

    # Print diagnostics
    print(f"\n=== Timestamp Comparison: {name1} vs {name2} ===")

    if missing_in_df1:
        print(f"\n⚠ Missing timestamps in {name1} (present in {name2}):")
        for t in missing_in_df1[:10]:
            print(" ", t)
        if len(missing_in_df1) > 10:
            print(f"  ... and {len(missing_in_df1)-10} more")

    if missing_in_df2:
        print(f"\n⚠ Missing timestamps in {name2} (present in {name1}):")
        for t in missing_in_df2[:10]:
            print(" ", t)
        if len(missing_in_df2) > 10:
            print(f"  ... and {len(missing_in_df2)-10} more")

    # Overlapping timestamps
    overlap = sorted(ts1.intersection(ts2))

    if overlap:
        print(f"\n✓ Overlapping date range:")
        print(f"  Start: {overlap[0]}")
        print(f"  End:   {overlap[-1]}")
        print(f"  Total overlapping timestamps: {len(overlap)}")
    else:
        print("\n❌ No overlapping timestamps found.")

    # If both sets match perfectly
    if not missing_in_df1 and not missing_in_df2:
        print(f"\n✓ Perfect timestamp alignment between {name1} and {name2}")

    # Perform inner join (only overlapping timestamps)
    final = df1.join(df2, how="inner")

    print(f"\n✓ Final merged shape: {final.shape}")
    return final
```

```python
final_df = merge_dataframes_wrt_time(wind_hourly, weather_hourly)
```
